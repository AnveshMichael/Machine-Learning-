```python
import numpy as np 
import pandas as pd
```

```python
nums =[45,48,49,50,51,52,47,54,46,53,50,49,55,44,52,48,51,47,56,43,50,54,46,52,49]
```

```python
arr=np.array(nums)
```

```python
arr
```

```python
arr.mean()
```

```python
mean= arr.mean()
std= arr.std()
```

```python
new = []
for i in range (len(nums)):
    scaler = (arr[i] - mean) / std
    new.append(scaler)
```

```python
scaled_arr = np.array(new)
```

```python
scaled_arr
```

```python
scaled_arr.mean()
```

```python
np.round(scaled_arr.mean())
```

```python
scaled_arr.std()
```
